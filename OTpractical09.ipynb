```python
from pulp import *

model = LpProblem("Integer_Programming", LpMaximize)

x = LpVariable("x", lowBound=0, cat="Integer")
y = LpVariable("y", lowBound=0, cat="Integer")

model += 5 * x + 4 * y

model += 6 * x + 4 * y <= 24
model += x + 2 * y <= 6

model.solve()

print("Status:", LpStatus[model.status])

print("Optimal Value of x =", value(x))
print("Optimal Value of y =", value(y))

print("Maximum Profit =", value(model.objective))




```
